# Tools, Resources & Prompts

- **Tools** are essentially functions which are designed to perform some "action"
- **Resources** are static (they don't perform any action unlike to tools)
- **Prompts** are templates


## Examples

Sr.No.  |   Item    |   Tool/Resource/Template
1   |   register_new_student    |   Tool
2   |   python_syllabus.pdf     |   Resource
3   |   generate_internship_offer_letter    |   Prompt
4   |   students.db |   Resource
5   |   welcome_whatsapp_msg   |   Prompt
6   |   send_onboarding_email   |   Tool
7   |   attendance_report.csv   |   Resource

Think of this instruction from the user:

"Read the privacy policy of the company, summarize it properly and then email me the summary."
