# Why do we need multiple mcp servers?

Consider the following example:

- We have created a chatbot for a school which can help answering users' questions.
- For this it needs different tools related to students, courses, attendance, calculations etc.
- Won't it make more organized structure if we put these tools in different servers like "student_server", "attendance_server" etc rather than dumping all the tools in one server?

# The issues

Earlier our client used `ClientSession` to connect to our one & only server using `server_params` which was as this:

```
server_params = StdioServerParameters(
    command= "python",
    args= ["server.py"]
)
```

Now we have three (or more) servers, so should we change `server_params`?

Or should we create multiple `ClientSession`?

If so then how our agent will manage multiple `ClientSession`?

What if tools from different servers are tobe called to answer one prompt? For example, what if the user asks - "How many students are there and what is 25 multiplied by 8?"

How to address all these problems?

**Solution: `ClientSessionGroup`**

# `ClientSessionGroup`

Its job is to:
- Connect to multiple servers
- Discover tools from all the servers
- Route tool call to the correct server automatically
- Present one, unified tool catalog to our client/agent
